```python
import pandas as pd
from google.cloud import bigquery
```

```python
client = bigquery.Client(project= "project-f78b761a-b634-4511-b83")

# Perform a query.
QUERY = (
    'SELECT text FROM `bigquery-public-data.hacker_news.full` LIMIT 1000')
query_job = client.query(QUERY)  # API request
rows = query_job.result()  # Waits for query to finish
```

```python
for row in rows:
    print(row.name)
```
